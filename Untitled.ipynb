```python
#5) factorial number
num = int(input("enter a number:"))
fact = 1
for i in range(num,0,-1):
    fact = fact*i
print(f"factorial:",fact)
```

```python
#8) reverse a number 12345 --> 54321
n= int(input("enter a num:"))
temp=n
rev=0
while(n>0):
    rem=n%10
    rev= (rev*10)+rem
    n=n//10
print(f"number:{n}\n reverse num:{rev}")
if(temp==rev):
    print(temp,"is a palindrome")
else:
    print(temp,"is not a palindrome")
```
