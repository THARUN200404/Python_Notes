```python
# write a program to print given number is prime or not

n = int(input("enter a number:"))
count = 0
for i in range(2,n,1):
    if(n%i==0):
        count=count+1
print("count:",count)
if count == 0:
    print(n,"is prime")
else: 
    print(n,"is not prime")


    
```

```python
num = 535
temp =num
rev=0
while(num>0):
    rem = num%10
    rev = (rev*10)+rem
    num=num//10
print(f"number:{temp}\nreversed num:{rev}")
if(temp==rev):
    print(temp, "is palindrome")
else:
    print(temp, "is not palindrome")
```

```python
ch=65
i=1
while(i<=5):
    for j in range(1,i+1,1):
        print(chr(ch),end=" ")
        ch+=1
    print()
    i+=1

#lower part

i=4
while(i>=1):
    for j in range(1,i+1,1):
        print(chr(ch),end=" ")
        ch+=1
    print()
    i-=1
```
