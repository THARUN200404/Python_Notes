```python
marks = int(input("enter percentage:"))
if(marks<0 and marks>100):
    print("invalid")
elif(marks >= 85):
    print("distinction")
elif (marks >= 60):
    print("first class")
elif(marks >= 50):  
    print("second class")
elif(marks>=35):
    print("pass")
else:
    print("fail")
```

```python
# write a program to print its equlater traingle ,isolated traingle,scalene traingle or not a traingle

a = float(input())
b = float(input())
c = float(input())
if (a==b==c):
    print("equaliteral traingle")
elif(a==b and b!=c):
    print("isolated traingle")
elif(a!=b and b!=c):
    print("scalen traingle")
elif(a+b<c or b+c<a or c+a<b):
    print("not a traingle")
     
```
