```python
#zero one traingle
n=4
for i in range(1,n+1):
    for j in range(1,i+1):
        if (i+j)%2==0:
            print(1,end=" ")
        else:
            print(0,end=" ")
    print()        
```

```python
#number increasing pyramid
n=4
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()    
        
```
