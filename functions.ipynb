```python
#create a function to print prime numbers between given range using all the methods

def prime():
    a=int(input("enter a number:"))
    b=int(input("enter a number:"))
    for n in range(a,b+1):
        if n%1==0 and n%n==0:
            print(n,"number is prime")
        else:
            print(n,"is  not prime")
prime()                
```

```python
lambda function 
a lambda function is a smal,anonymous(nameless)
function in python

* it drfine using keyword lambda instead of def
* it can take any number of arguements but must contain only one expression
* expresion is automatically returned(no need to use return)


why to use lambda
* useful for short ,simple operation
* saves code(one-liner)
* often used with map(),filter(),reduce(), or as inline funcion

why not use lambda?
* for complex logic
* when readility is more important then shortness.
```
